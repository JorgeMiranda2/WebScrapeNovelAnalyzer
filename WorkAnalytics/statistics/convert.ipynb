```python
import tkinter as tk
from tkinter import filedialog
```

```python
root = tk.Tk()
root.withdraw()  # Ocultar la ventana principal

# Directorio inicial para la búsqueda del archivo CSV
initial_dir = '../Data/'

# Abrir un cuadro de diálogo para seleccionar el archivo
file_path = filedialog.askopenfilename(initialdir=initial_dir, filetypes=[("CSV files", "*.csv")])

```

```python
# Verificar si se seleccionó un archivo
if file_path:
    # Leer el archivo CSV seleccionado usando pandas
    dataframe = pd.read_csv(file_path, delimiter=';')
    
    # Hacer lo que necesites con el DataFrame (por ejemplo, imprimir las primeras filas)
    print(dataframe.head().isna().sum().sort_values(ascending=False))
else:
    print("Ningún archivo seleccionado.")
```

```python
def categorys_graphic(cat):
    from math import ceil
    rows  = ceil(cat.shape[1]/2)
    #Definir grafico
    f, ax = plt.subplots(nrows=rows, ncols = 2, figsize=(16,rows*16))
    ax = ax.flat

    # crear bucle

    for index, variable in enumerate(cat):
        cat[variable].value_counts().plot.barh(ax = ax[index])
        ax[index].set_title(variable, fontsize=12, fontweight='bold')
        ax[index].tick_params(labelsize=12)
```
